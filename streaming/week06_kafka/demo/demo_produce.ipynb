# Demo: Kafka Producer

Minimal working example — sends one event to Redpanda.

No TODOs. Just run each cell top-to-bottom and observe what happens.

**Before you run:** make sure the Codespace is up — `rpk cluster info` should print broker info in a terminal.

```python
from confluent_kafka import Producer
import json, time

BROKER = 'redpanda:29092'   # Redpanda's internal listener inside the docker network

producer = Producer({'bootstrap.servers': BROKER})
print(f'Producer connected to {BROKER}')
```

```python
# Define the event
topic = 'strom'
key   = 'haus_a'                         # determines the partition
value = json.dumps({                     # the actual payload
    'haus':      'haus_a',
    'wert':      42.5,
    'einheit':   'kWh',
    'timestamp': time.time(),
})

print(f'Topic: {topic}')
print(f'Key:   {key}')
print(f'Value: {value}')
```

```python
def on_delivery(err, msg):
    if err:
        print(f'Failed: {err}')
    else:
        print(f'Delivered -> topic={msg.topic()} partition={msg.partition()} offset={msg.offset()}')

producer.produce(topic, key=key.encode(), value=value.encode(), callback=on_delivery)
producer.flush()   # blocks until delivery is confirmed
print('Done.')
```

**Check it worked:**

1. Open the **Redpanda Console** (port `8080` in the *PORTS* tab).
2. Click *Topics* → *strom* → *Messages*.
3. You should see your event with key `haus_a` and the JSON value.
